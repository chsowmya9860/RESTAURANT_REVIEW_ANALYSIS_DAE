Filling the NaN values with specific value

```python
import pandas as pd
df = pd.read_excel('/content/drive/MyDrive/dae_dataset.xlsx',sheet_name = 'Sheet1')
df["Ambience_Rating"] = df["Ambience_Rating"].fillna("Not rated")
df["Service_Rating"] = df["Service_Rating"].fillna("Not rated")
df["Review_Text"] = df["Review_Text"].fillna("Not reviewed")
df["Feedback"] = df["Feedback"].fillna("No feedback")
print("Dataset after  filling missing values")
print(df)
```

Adding a new column(Category)

```python
import pandas as pd
df = pd.read_excel('/content/drive/MyDrive/dae_dataset.xlsx',sheet_name = 'Sheet1')
print("Printing After adding a new column")
df['Category'] = df['Overall_Rating'].apply(
      lambda x:'Most Popular' if x==5
      else 'Popular' if x==4
      else 'Not Popular')
print(df)
```

Performing the operations on the dataset using aggregate functions

```python
import pandas as pd
df = pd.read_excel('/content/drive/MyDrive/dae_dataset.xlsx',sheet_name = 'Sheet1')
print("Finding the average of feedbacks based on rating")
avg_feedback = df.groupby("Feedback")["Overall_Rating"].mean()
print(avg_feedback)
print("Finding the count of feedbacks based on rating")
feeback_count = df.groupby("Feedback")["Overall_Rating"].count()
print(feeback_count)
print("Finding the maximum of feedbacks based on rating")
max_feedback = df.groupby("Feedback")["Overall_Rating"].max()
print(max_feedback)
print("Finding the minimum of feedbacks based on rating")
min_feedback = df.groupby("Feedback")["Overall_Rating"].min()
print(min_feedback)
```

bar graph

```python
import matplotlib.pyplot as plt
restaurant_rating = df.groupby('Restaurant_Name')['Ambience_Rating'].sum()
plt.figure(figsize=(7,5))
plt.bar(restaurant_rating.index, restaurant_rating.values,color = "steelblue")
plt.xlabel('Restaurant Name')
plt.ylabel('Ambience Rating')
plt.title('Ambience Rating by Restaurant')
plt.tight_layout()
plt.savefig("bar_graph.png",dpi = 120)
plt.close()
```

pie chart

```python
feedback_count = df['Feedback'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(feedback_count.values,labels = feedback_count.index, autopct = "%1.1f%%", startangle = 90)
plt.title('Feedback Distribution')
plt.tight_layout()
plt.savefig("pie_chart.png",dpi = 120)
plt.close()
```

box_plot

```python
feedback_values = df["Feedback"].unique()
data_by_feedback = [df[df["Feedback"] == feedback]["Overall_Rating"].dropna() for feedback in feedback_values]
plt.figure(figsize=(8,5))
plt.boxplot(data_by_feedback, tick_labels = feedback_values)
plt.xlabel("Feedback")
plt.ylabel("Overall Rating")
plt.title("Overall Rating Distribution by Feedback")
plt.tight_layout()
plt.savefig("box_plot.png",dpi = 120)
plt.close()
```

histogram

```python
plt.figure(figsize=(7,5))
plt.hist(df["Food_Rating"].dropna(), bins = 8, color = "mediumseagreen", edgecolor = "black")
plt.xlabel("Food_Rating")
plt.ylabel("Frequency")
plt.title("Distribution of food ratings")
plt.tight_layout()
plt.savefig("histogram.png",dpi = 120)
plt.close()
```

Line chart

```python
fig, axes = plt.subplots(2,1,figsize=(8,8),sharex = True)
axes[0].plot(df.index,df["Food_Rating"],marker="o",color="royalblue",label="Food Rating")
axes[0].set_title("Food Rating Trend")
axes[0].set_ylabel("Food Rating")
axes[0].legend()
axes[0].grid(True,alpha=0.3)
axes[1].plot(df.index,df["Service_Rating"],marker="s",color="darkorange",label="Service Rating")
axes[1].set_title("Service Rating Trend")
axes[1].set_xlabel("Record Index")
axes[1].set_ylabel("Service Rating")
axes[1].legend()
axes[1].grid(True,alpha=0.3)
plt.tight_layout()
plt.savefig("line_chart_subplots.png",dpi = 120)
plt.close()
```
